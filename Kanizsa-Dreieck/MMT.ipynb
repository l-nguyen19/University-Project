```python
# KANIZSA-DREIECK

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

fig, ax = plt.subplots()

# Achsen definieren
plt.xlim(-5,15)
plt.ylim(-5,15)

x = 0
y = 0

# Länge
L = 10

# Koordinaten Mittelpunkt Kreis 1
c1 = (0,0)

# Gradmaß -> Bogenmaß
gradmaß = 60
bogenmaß = (2*np.pi*gradmaß)/360

# Berechnung der neuen Punkte mit Winkel
x2 = x +(L*np.cos(bogenmaß))
y2 = y +(L*np.sin(bogenmaß))

# Koordinaten Mittelpunkt Kreis 2
c2 = (x2,y2)

# Koordinaten Mittelpunkt Kreis 3
c3 = (10,0)

# Zeichnen der Kreise
cir1 = plt.Circle((c1), 1.5, color='black')
ax.add_patch(cir1)
cir2 = plt.Circle((c2), 1.5, color='black')
ax.add_patch(cir2)
cir3 = plt.Circle((c3), 1.5, color='black')
ax.add_patch(cir3)

# Berechnung der Liniensegmente
winkel = 300
t2x2 = 0 +(L*np.cos((2*np.pi*winkel)/360))
t2y2 = 6 +(L*np.sin((2*np.pi*winkel)/360))

# Liniensegmente(Dreieck)
triangle2 = np.array([(0,6), (t2x2,t2y2), (10,6)])
p2 = Polygon(triangle2, facecolor = "white", edgecolor = "black")
ax = plt.gca()
ax.add_patch(p2)

# weißes Dreieck
triangle1 = np.array([c1, c2, c3])
p1 = Polygon(triangle1, closed=True, color='white')
ax = plt.gca()
ax.add_patch(p1)

#plot
ax.set_facecolor("white")
ax.set_aspect('equal', adjustable='box')
plt.title( 'Kanizsa-Dreieck' )
plt.show()
```

```python
# WINKEL VERÄNDERBAR

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

fig, ax = plt.subplots()

# Achsen definieren
plt.xlim(-8,10)
plt.ylim(-8,10)

# Länge der neu berechneten Seite
L = 8

# Koordinaten Mittelpunkt Kreis 1
c1 = (0,0)

# Gradmaß -> Bogenmaß
gradmaß = float(input("Winkel in Gradmaß eingeben:"))
bogenmaß = (2*np.pi*gradmaß)/360

# Berechnung der neuen Punkte mit Winkel
x2 = (L*np.cos(bogenmaß))
y2 = (L*np.sin(bogenmaß))

# Koordinaten Mittelpunkt Kreis 2
c2 = (x2,y2)

# Koordinaten Mittelpunkt Kreis 3
c3 = (8,0)

# Zeichnen der Kreise
cir1 = plt.Circle((c1), 1.5, color='black')
ax.add_patch(cir1)
cir2 = plt.Circle((c2), 1.5, color='black')
ax.add_patch(cir2)
cir3 = plt.Circle((c3), 1.5, color='black')
ax.add_patch(cir3)

# weißes Dreieck definieren
triangle1 = np.array([c1, c2, c3])
p1 = Polygon(triangle1, closed=True, color='white')
ax = plt.gca()
ax.add_patch(p1)

# plot
ax.set_aspect('equal', adjustable='box')
plt.title( 'Winkel veränderbar' )
plt.show()
```

```python
# POSITION DER KREISE VERÄNDERBAR

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

fig, ax = plt.subplots()

# Achsen definieren
plt.xlim(-10,10)
plt.ylim(-10,10)

# Eingabe Mittelpunkte der Kreise
x1 = float(input("x-Koordinate des ersten Kreises:"))
y1 = float(input("y-Koordinate des ersten Kreises:"))

x2 = float(input("x-Koordinate des zweiten Kreises:"))
y2 = float(input("y-Koordinate des zweiten Kreises:"))

x3 = float(input("x-Koordinate des dritten Kreises:"))
y3 = float(input("y-Koordinate des dritten Kreises:"))

# Mittelpunkt der Kreise
c1 = (x1,y1)
c2 = (x2,y2)
c3 = (x3,y3)

# Kreise definieren
cir1 = plt.Circle((c1), 1.5, color='black')
ax.add_patch(cir1)
cir2 = plt.Circle((c2), 1.5, color='black')
ax.add_patch(cir2)
cir3 = plt.Circle((c3), 1.5, color='black')
ax.add_patch(cir3)

# weißes Dreieck definieren
triangle1 = np.array([c1, c2, c3])
p1 = Polygon(triangle1, closed=True, color='white')
ax = plt.gca()
ax.add_patch(p1)

# plot
ax.set_aspect('equal', adjustable='box')
plt.title( 'Position der Kreise veränderbar' )
plt.show()
```
